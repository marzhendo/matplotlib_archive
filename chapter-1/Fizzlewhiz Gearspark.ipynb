```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('gearsparks_emporium_weekly_sales.csv')

print('data siap dipakai')
```

```python
# Plot penjualan senjata per hari setiap minggu
plt.figure(figsize=(10, 6))
plt.plot(df['Day'], df['Weapon'], marker='o', label='Weapon Sales')
plt.title('Weekly Weapon Sales at Gearspark Emporium')
plt.xlabel('Day of the Week')
plt.ylabel('Number of Weapons Sold')
plt.grid()
plt.legend()
plt.show()
```
