```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
games = [1, 2, 3, 4, 5]

# Points for the NBA teams
bulls_points = [89, 104, 132, 203, 115]
lakers_points = [102, 110, 95, 108, 115]
celtics_points = [98, 105, 100, 112, 107]
warriors_points = [110, 99, 105, 120, 115]

plt.figure(figsize=(10, 6))
plt.plot(games, bulls_points, marker='o', label='Bulls', color='red', linestyle='--')
plt.plot(games, lakers_points, marker='s', label='Lakers', color='purple', linestyle='-.')
plt.plot(games, celtics_points, marker='^', label='Celtics', color='green', linestyle=':')
plt.plot(games, warriors_points, marker='d', label='Warriors', color='blue', linestyle='-')

plt.xlabel('Games')
plt.ylabel('Points')
plt.title('NBA Team Points Over Time')
plt.legend()
plt.grid(True)
plt.show()
```
