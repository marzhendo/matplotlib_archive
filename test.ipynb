```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Data
year = np.array([2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025])
revenue_action = np.array([4.8, 5.3, 2.1, 3.4, 5.9, 6.5, 7.2, 8.1])
revenue_comedy = np.array([3.5, 3.8, 1.5, 2.2, 4.1, 4.6, 5.0, 5.5]) # Data baru!

# Plotting
plt.plot(year, revenue_action, marker='o', linestyle='--', color='red', label='Action Movies')
plt.plot(year, revenue_comedy, marker='s', linestyle='-', color='blue', label='Comedy Movies') # Baris baru!

plt.title('Action vs Comedy Box Office Revenue (2018-2025)')
plt.xlabel('Year')
plt.ylabel('Revenue (billions USD)')
plt.legend()
plt.show()
```
