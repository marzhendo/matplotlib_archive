```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
days = np.array(["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"])
steps = np.array([3000, 4500, 7000, 8000, 6500, 9000, 4000])

plt.figure()
plt.plot(days, steps, marker='o')
plt.title("Langkah Kaki Mingguan")
plt.xlabel("Hari")
plt.ylabel("Jumlah Langkah")
plt.show()
```
