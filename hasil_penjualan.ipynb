```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data Bulan
months = np.array(['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])

# Data Penjualan (dalam Juta Rp)
sales_2023 = np.array([15, 18, 20, 22, 25, 28, 30, 28, 35, 40, 45, 50])
sales_2024 = np.array([55, 58, 60, 63, 65, 70, 72, 75, 78, 85, 95, 100])
x = np.arange(len(months))  # posisi untuk label bulan
width = 0.4
```

```python
# Visualisasi Data
plt.plot(months, sales_2023, marker='o', label='Penjualan 2023', color='blue')
plt.plot(months, sales_2024, marker='o', label='Penjualan 2024', color='orange', linestyle='--')
plt.title('Hasil Penjualan Tahunan (2023 vs 2024)')
plt.xlabel('Bulan')
plt.ylabel('Penjualan (Juta Rp)')
plt.legend()

plt.show()
```

```python
# Dalam bentuk bar chart
plt.bar(x - width/2, sales_2023,width=width, alpha=0.6, label='Penjualan 2023', color='blue')
plt.bar(x + width/2, sales_2024, width=width, alpha=0.6, label='Penjualan 2024', color='orange')
plt.title('Hasil Penjualan Tahunan (2023 vs 2024)')
plt.xticks(x, months)
plt.xlabel('Bulan')
plt.ylabel('Penjualan (Juta Rp)')
plt.legend()
plt.show()
```
