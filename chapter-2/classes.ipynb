```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.DataFrame({
    "Exam": ["CS", "Physics", "Calculus", "Anatomy"],
    "Alice": [90, 85, 80, 88],
    "Bob": [85, 87, 82, 90],
    "Charlie": [88, 90, 84, 86],
    "Diana": [92, 88, 85, 89]
})

print("DataFrame siap digunakan!")
```

```python
# Buat 4 Subplots (4 baris, 1 kolom)
fig, axes = plt.subplots(nrows=4, ncols=1, figsize=(8, 10))
# Plot untuk setiap siswa
axes[0].bar(df["Exam"], df["Alice"], color='blue')
axes[0].set_title("Alice's Scores")
axes[1].bar(df["Exam"], df["Bob"], color='orange')
axes[1].set_title("Bob's Scores")
axes[2].bar(df["Exam"], df["Charlie"], color='green')
axes[2].set_title("Charlie's Scores")
axes[3].bar(df["Exam"], df["Diana"], color='red')
axes[3].set_title("Diana's Scores")
# Atur layout agar tidak tumpang tindih
plt.tight_layout()
# Tampilkan plot
plt.show()
```

```python
# Sub plot dengan 2 baris dan 2 kolom dengan tambahan parameter sharex=True
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(10, 8), sharex=True)
# Plot untuk setiap siswa
axes[0, 0].bar(df["Exam"], df["Alice"], color='blue')
axes[0, 0].set_title("Alice's Scores")
axes[0, 1].bar(df["Exam"], df["Bob"], color='orange')
axes[0, 1].set_title("Bob's Scores")
axes[1, 0].bar(df["Exam"], df["Charlie"], color='green')
axes[1, 0].set_title("Charlie's Scores")
axes[1, 1].bar(df["Exam"], df["Diana"], color='red')
axes[1, 1].set_title("Diana's Scores")
# Atur layout agar tidak tumpang tindih
plt.tight_layout()
# Tampilkan plot
plt.show()
```

```python
# Dataset Penjualan baju
df_baju = pd.DataFrame({
    'Kuartal': ['Q1', 'Q2', 'Q3', 'Q4'],
    'Kaos': [120, 150, 145, 130],
    'Kemeja': [90, 110, 100, 95],
    'Celana': [100, 105, 110, 115],
    'Jaket': [30, 40, 50, 160]
})

print(df_baju)
```

```python
# Buat subplot untuk penjualan baju dengan 2 baris dan 2 kolom
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(10, 8), sharex=True)
# Plot untuk Kaos
axes[0, 0].bar(df_baju['Kuartal'], df_baju['Kaos'], color='blue')
axes[0, 0].set_title("Penjualan Kaos")
# Plot untuk Kemeja
axes[0, 1].bar(df_baju['Kuartal'], df_baju['Kemeja'], color='orange')
axes[0, 1].set_title("Penjualan Kemeja")
# Plot untuk Celana
axes[1, 0].bar(df_baju['Kuartal'], df_baju['Celana'], color='green')
axes[1, 0].set_title("Penjualan Celana")
# Plot untuk Jaket
axes[1, 1].bar(df_baju['Kuartal'], df_baju['Jaket'], color='red')
axes[1, 1].set_title("Penjualan Jaket")
# Atur layout agar tidak tumpang tindih
plt.tight_layout()
# Tampilkan plot
plt.show()
```

```python
# Pake looping untuk plot
items = ['Kaos', 'Kemeja', 'Celana', 'Jaket']
colors = ['blue', 'orange', 'green', 'red']
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(10, 8), sharex=True)
# Pake .flatten() untuk meratakan array axes agar bisa di-looping
# for i, item in enumerate(items):
#     axes.flatten()[i].bar(df_baju['Kuartal'], df_baju[item], color=colors[i])
#     axes.flatten()[i].set_title(f"Penjualan {item}")
for ax, items, color in zip(axes.flatten(), items, colors):
    ax.bar(df_baju['Kuartal'], df_baju[items], color=color)
    ax.set_title(f"Penjualan {items}")
# Atur layout agar tidak tumpang tindih
plt.tight_layout()
# Tampilkan plot
plt.show()
```
