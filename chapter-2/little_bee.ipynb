```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Data dummy untuk Mini-Bees 🐝
data = {
    'state': ['California', 'Texas', 'Florida', 'New York', 'Georgia'] * 4,
    'year': [2019]*5 + [2020]*5 + [2021]*5 + [2022]*5,
    'num_colonies': [15000, 12000, 8000, 6000, 7000, 16000, 12500, 8500, 6200, 7200, 15500, 11800, 8200, 6100, 7100, 14500, 11500, 8100, 6300, 7300],
    'percent_lost': [15, 12, 10, 8, 9, 18, 14, 11, 9, 10, 20, 15, 12, 10, 11, 22, 18, 14, 11, 12],
    'varroa_mites': [5.2, 4.1, 3.5, 2.8, 3.0, 6.5, 5.0, 4.2, 3.1, 3.5, 7.8, 5.8, 4.8, 3.5, 4.0, 8.5, 6.2, 5.1, 3.8, 4.5]
}

df_bees = pd.DataFrame(data)
print(df_bees.head())
```

```python
# Buat bar chart untuk menunjukan rata-rata presentase kehilangan untuk setiap negara bagian
avg_lost = df_bees.groupby('state')['percent_lost'].mean().reset_index()
plt.figure(figsize=(10, 6))
plt.bar(avg_lost['state'], avg_lost['percent_lost'], color='orange')
plt.title('Rata-rata Presentase Kehilangan Koloni Lebah per Negara Bagian')
plt.xlabel('Negara Bagian')
plt.ylabel('Rata-rata Presentase Kehilangan (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Buat bar chart untuk total jumlah koloni yang dimiliki setiap negara bagian
total_colonies = df_bees.groupby('state')['num_colonies'].sum().reset_index()
plt.figure(figsize=(10, 6))
plt.bar(total_colonies['state'], total_colonies['num_colonies'], color='green')
plt.title('Total Jumlah Koloni Lebah per Negara Bagian')
plt.xlabel('Negara Bagian')
plt.ylabel('Total Jumlah Koloni')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Buat line plot untuk menunjukan apakah rata-rata persentase kehilangan (percent_lost) cenderung naik atau turun dari tahun ke tahun?
bees_per_year = df_bees.groupby('year')['percent_lost'].mean().reset_index()
plt.figure(figsize=(12, 8))
plt.plot(bees_per_year['year'], bees_per_year['percent_lost'], marker='o', color='blue')
plt.title('Tren Rata-rata Persentase Kehilangan Koloni Lebah dari Tahun ke Tahun')
plt.xlabel('Tahun')
plt.ylabel('Rata-rata Persentase Kehilangan (%)')
plt.show()
```

```python
# Buat scatter plot untuk melihat Apakah hama Varroa Mites (Tungau) benar-benar menyebabkan hilangnya koloni?
plt.figure(figsize=(10, 6))
plt.scatter(df_bees['varroa_mites'], df_bees['percent_lost'], color='red')
plt.title('Hubungan antara Varroa Mites dan Persentase Kehilangan Koloni Lebah')
plt.xlabel('Jumlah Varroa Mites')
plt.ylabel('Persentase Kehilangan (%)')
plt.show()
```
