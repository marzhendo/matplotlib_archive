```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('gas_electricity_rent_expenses.csv')
df.head()
```

```python
# Filter data dulu menjadi DataFrame utuh
electric_df = df[df['Category'] == 'Electric']
gas_df = df[df['Category'] == 'Gas']

fig, ax = plt.subplots(figsize=(10, 6))

# Plot langsung dari DataFrame hasil filter
# X = electric_df['Month'], Y = electric_df['Amount_USD']
ax.plot(electric_df['Month'], electric_df['Amount_USD'], marker='o', label='Electricity', color='blue')

# Lakukan hal yang sama untuk Gas
ax.plot(gas_df['Month'], gas_df['Amount_USD'], marker='s', label='Gas', color='red')

ax.set_title('Monthly Expenses for Electricity and Gas')
ax.set_xlabel('Month')
ax.set_ylabel('Amount (USD)')
ax.legend()
plt.show()
```
