```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.DataFrame({
    'Week': ['Week 1', 'Week 2', 'Week 3', 'Week 4', 'Week 5'],
    'The_Housemaid_Sales': [31.01, 29.82, 22.12, 15.67, 12.63],
    'Marty_Supreme_Sales': [11.55, 30.75, 19.05, 11.67, 8.57]
})
```

```python
# OOP(Object-Oriented Plotting)
fig, ax = plt.subplots()
fig.set_size_inches(10, 6)
ax.plot(df['Week'], df['The_Housemaid_Sales'], label='The Housemaid Sales', marker='o')
ax.plot(df['Week'], df['Marty_Supreme_Sales'], label='Marty Supreme Sales', marker='o')
ax.set_title('Box Office Sales Over 5 Weeks')
ax.set_xlabel('Week')
ax.set_ylabel('Sales (in millions)')
ax.legend()
plt.show()
```
