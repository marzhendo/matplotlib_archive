```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.DataFrame({
  'aquarium': ['Monterrey Bay', 'Georgia Aquarium', 'Shedd Aquarium', 'National Aquarium'],
  'shark': [5, 4, 4, 2],
  'dolphin': [8, 12, 10, 5],
  'octopus': [7, 20, 8, 4],
  'sea turtle': [12, 5, 6, 3],
  'jellyfish': [30, 25, 20, 15]
})

df.head()
```

```python
# Visualisasi Hiu di setiap akuarium
plt.figure(figsize=(10, 6))
plt.bar(df['aquarium'], df['shark'], color='blue')
plt.xlabel('Aquarium')
plt.xticks(rotation=45)
plt.ylabel('Number of Sharks')
plt.title('Number of Sharks in Each Aquarium')
plt.show()
```
