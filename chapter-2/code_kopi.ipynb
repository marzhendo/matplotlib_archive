```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Hari': ['Senin', 'Selasa', 'Rabu', 'Kamis', 'Jumat', 'Sabtu', 'Minggu'],
    'Kopi_Panas': [50, 45, 55, 60, 58, 80, 75],
    'Es_Kopi': [30, 35, 40, 55, 70, 95, 90]
}

df = pd.DataFrame(data)
```

```python
# OOP(Object-Oriented Plotting)
fig, ax = plt.subplots()
fig.set_size_inches(10, 6)
# Plot data kopi panas
ax.plot(df['Hari'], df['Kopi_Panas'], label='Kopi Panas', marker='o')
# Plot data es kopi
ax.plot(df['Hari'], df['Es_Kopi'], label='Es Kopi', marker='o')
ax.set_title('Penjualan Kopi Panas dan Es Kopi Selama Seminggu')
ax.set_xlabel('Hari')
ax.set_ylabel('Jumlah Penjualan')
ax.legend()
plt.show()
```

```python
# Pakai sublots
fig, ax = plt.subplots(nrows=2)
# Plot data kopi panas
ax[0].plot(df['Hari'], df['Kopi_Panas'], label='Kopi Panas', marker='o')
ax[0].set_title('Penjualan Kopi Panas Selama Seminggu')
ax[0].set_xlabel('Hari')
ax[0].set_ylabel('Jumlah Penjualan')
ax[0].legend()
# Plot data es kopi
ax[1].plot(df['Hari'], df['Es_Kopi'], label='Es Kopi', marker='o', color='orange')
ax[1].set_title('Penjualan Es Kopi Selama Seminggu')
ax[1].set_xlabel('Hari')
ax[1].set_ylabel('Jumlah Penjualan')
ax[1].legend()
plt.tight_layout()
plt.show()
```
