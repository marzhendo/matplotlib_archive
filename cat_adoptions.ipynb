```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({
  'month': ['Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
  'adoptions': [45000, 52000, 48000, 43000, 38000, 41000]
})
```

```python
plt.figure(figsize=(8, 5))
plt.plot(df['month'], df['adoptions'], marker='o', linestyle='-', color='b')
plt.title('Cat Adoptions Over Months')
plt.xlabel('Month')
plt.ylabel('Number of Adoptions')
plt.grid(True)
plt.show()
```
