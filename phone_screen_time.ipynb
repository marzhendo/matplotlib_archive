```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
days = np.array(["Senin", "Selasa", "Rabu", "Kamis", "Jumat", "Sabtu", "Minggu"])
screen_time = np.array([342, 298, 310, 275, 290, 400, 185])
x = np.arange(len(days))
plt.plot(x, screen_time, marker="o", linestyle="-", color="b")
plt.xticks(x, days)
plt.xlabel("Hari")
plt.ylabel("Waktu Layar (menit)")
plt.title("Waktu Layar Harian")
plt.grid()
plt.show()
```

```python
# Hitung Rata-rata Waktu Layar
avg_screen_time = np.average(screen_time)
print(f"Rata-rata Waktu Layar Harian: {avg_screen_time:.2f} menit")
```
